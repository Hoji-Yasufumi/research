# **ローパスフィルターをかける関数**

```python
import scipy # scipyライブラリをインポート
import scipy.signal

from scipy import signal
!pip install --upgrade scipy

def butter_lowpass(lowcut, fs, order=4):
    '''バターワースローパスフィルタを設計する関数
    '''
    nyq = 0.5 * fs
    low = lowcut / nyq
    b, a = signal.butter(order, low, btype='low')
    return b, a


def butter_lowpass_filter(x, lowcut, fs, order=4):
    '''データにローパスフィルタをかける関数
    '''
    b, a = butter_lowpass(lowcut, fs, order=order)
    y = signal.filtfilt(b, a, x)
    return y

# 高速フーリエ変換
def calc_amp(data, fs):
    '''フーリエ変換して振幅スペクトルを計算する関数
    '''
    N = len(data)
    window = scipy.signal.windows.hann(N)
    F = np.fft.fft(data * window)
    freq = np.fft.fftfreq(N, d=1/fs) # 周波数スケール
    F = F / (N / 2) # フーリエ変換の結果を正規化
    F = F * (N / sum(window)) # 窓関数による振幅減少を補正する
    Amp = np.abs(F) # 振幅スペクトル
    return Amp, freq

N = 90 #サンプル数
fs = 30 #サンプリング周波数

#ローパスフィルターをかける
def lowpass(pressesAdjustAve, cuttingTimes5000from8000):

    x=pressesAdjustAve
    t=cuttingTimes5000from8000


    Amp, freq = calc_amp(x, fs)
    fig, ax = plt.subplots()
    ax.plot(freq[:N//2], Amp[:N//2])
    ax.set_xlabel("Frequency [Hz]")
    ax.set_ylabel("Amplitude")
    ax.grid()
    plt.show()


    y = butter_lowpass_filter(x, 4, fs, order=4)
    """
    fig, ax = plt.subplots()
    ax.plot(t, x, label='raw data')
    ax.plot(t, y, label='filtered')
    ax.set_xlabel("Time [s]")
    ax.set_ylabel("Signal")
    #ax.set_xlim([0, 0.1])
    ax.grid()
    plt.legend(loc='best')
    #plt.show()
    """

    return y

```

```python
import numpy as np
from sklearn.model_selection import StratifiedKFold
from sklearn.datasets import make_classification
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# ==========================
# 1. データセットの準備
# ==========================
# 今回はサンプルとして、make_classificationで2クラス分類用の合成データを作成します
# n_samples: サンプル数
# n_features: 特徴量の次元
# n_informative: 分類に有効な特徴量の数
# random_state: 乱数の種
X, y = make_classification(n_samples=1000,
                           n_features=10,
                           n_informative=2,
                           n_redundant=0,
                           n_classes=2,
                           random_state=42)

print(X)
print(y)

# ==========================
# 2. StratifiedKFoldの準備
# ==========================
# k=5の交差検証を行う (5分割)
# shuffle=True で各エポックごとにデータをシャッフル
# random_state=42 で再現性のある分割
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# ==========================
# 3. 交差検証の実行
# ==========================
accuracies = []

for fold_idx, (train_index, test_index) in enumerate(skf.split(X, y), start=1):
    # トレーニングデータとテストデータを分割
    X_train, X_test = X[train_index], X[test_index]
    y_train, y_test = y[train_index], y[test_index]

    # 分割後のクラスの分布を確認 (任意)
    # np.bincountはクラスのカウントを返す
    print(f"Fold {fold_idx}")
    print("  Train クラス分布:", np.bincount(y_train))
    print("  Test  クラス分布:", np.bincount(y_test))

    # ==========================
    # 4. モデルの学習
    # ==========================
    model = LogisticRegression()
    model.fit(X_train, y_train)

    # テストデータで予測
    y_pred = model.predict(X_test)

    # ==========================
    # 5. 評価指標の計算
    # ==========================
    acc = accuracy_score(y_test, y_pred)
    accuracies.append(acc)
    print(f"  Accuracy: {acc:.4f}\n")

print("各FoldのAccuracy:", np.round(accuracies, 4))
print("平均Accuracy:", np.round(np.mean(accuracies), 4))
```
